```python
# importando a biblioteca pandas e os arquivos Csv
import pandas as pd
import numpy as np
import numpy as np
df_clientes = pd.read_csv('clientes.csv', sep=';')
df_itens_pedidos = pd.read_csv('Itens_pedidos.csv', sep=';')
df_pedidos = pd.read_csv('Pedidos.csv',sep=';')
df_produtos = pd.read_csv('Produtos.csv',sep=';')
df_vendedores = pd.read_csv('Vendedores.csv',sep=';')
display(df_clientes)
display(df_itens_pedidos)
display(df_pedidos)
display(df_produtos)
display(df_vendedores)
```

```python
# Nulos e duplicatas vendedores
display(df_vendedores.isnull().sum())
display(df_clientes.duplicated().sum())
```

```python
# Padronizando strings 
display(df_pedidos)
df_pedidos["status"]=df_pedidos["status"].str.replace("Em separação","Em processamento")
display(df_pedidos)
```

```python

# Conversão clientes
display(df_clientes.info())
df_clientes["data_cadastro"]=pd.to_datetime(df_clientes["data_cadastro"],dayfirst=True)
display(df_clientes.info())
```

```python
# Conversão itens pedidos
display(df_itens_pedidos.info())
df_itens_pedidos["quantidade"]=(
    df_itens_pedidos["quantidade"].astype("Int64"))

df_itens_pedidos['preco_unitario'] = pd.to_numeric(
    df_itens_pedidos['preco_unitario'].str.replace(',', '.', regex=False),
    errors='raise'
)

df_itens_pedidos['desconto_pct'] = pd.to_numeric(
    df_itens_pedidos['desconto_pct']
        .str.replace(',', '.', regex=False)
        .replace('sem desconto', '0'),
    errors='raise'
)


df_itens_pedidos.info()

display(df_itens_pedidos.dtypes)
display(df_itens_pedidos.isnull().sum())
```

```python
# Conversão pedidos
df_pedidos["data_pedido"]=pd.to_datetime(df_pedidos["data_pedido"],dayfirst=True)
display(df_pedidos.info())
```

```python
# Conversão produtos
display(df_produtos.info())
df_produtos['produto_id'] = pd.to_numeric(
    df_itens_pedidos['produto_id'],
    errors='raise'
)

df_produtos['custo_unitario']=pd.to_numeric(
    df_produtos['custo_unitario'].str.replace(',','.',regex=False),
    errors='raise'

)

df_produtos['preco_lista']=pd.to_numeric(
    df_produtos['preco_lista'].str.replace(',','.',regex=False),
    errors='raise'

)
display(df_produtos.info())
```

```python
# juntando tabelas 
# Junta df_itens_pedidos usando pedido_id

df_analise=df_itens_pedidos.merge(
                    df_pedidos,
                    on='pedido_id',
                    how='inner')

df_analise=df_analise.merge(
                    df_produtos,
                    on='produto_id',
                    how='left')

df_analise=df_analise.merge(
    df_clientes,
    on='cliente_id',
    how='left'
      )

df_analise=df_analise.merge(
    df_vendedores,
    on='vendedor_id',
    how='left'
)
display(df_analise)
```

```python
# Pedidos de janeiro e outubro
display(df_pedidos.info())
janeiro=df_pedidos[df_pedidos['data_pedido'].dt.month==1]
outubro=df_pedidos[df_pedidos['data_pedido'].dt.month==10]

df_resultado=pd.concat([janeiro,outubro])
display(df_resultado)

# Mês que teve mais pedidos e quantos pedidos existem em cada um.
df_analise=df_resultado.groupby(df_resultado['data_pedido'].dt.month).size()
display(df_resultado)

# Analisar os status dos pedidos
status_pedidos=df_resultado['status'].value_counts()
display(status_pedidos)

# Quais vendedores recebem mais pedidos
analise_vendedores=df_resultado.groupby('vendedor_id').size()
display(analise_vendedores)
```

```python

df_analise=df_itens_pedidos.merge(
    df_produtos,
    on='produto_id',
    how='left'
)

display(df_analise)

# Quantidade total vendida por categoria.
# Qreço médio dos itens por categoria.
categoria_analise=df_analise.groupby('categoria').agg(
    qtd_categoria=('quantidade','sum'),preco_medio=('preco_unitario','mean') )
display(categoria_analise)

# Categoria Com o maior faturamento.
df_analise['Faturamento']=df_analise['quantidade']*df_analise['preco_unitario']
df_analise=df_analise.groupby('categoria').agg(total_faturamento=('Faturamento','sum'))
df_analise=df_analise.sort_values(by='total_faturamento',ascending=False)
display(df_analise)

# Maior faturamento e menor faturamento
display(df_analise.head(1))
display(df_analise.tail(1))
```

```python
# Para resumo mensal
# Juntar itens dos pedidos com as datas e informações dos pedidos
df_completo = df_itens_pedidos.merge(
    df_pedidos,
    on='pedido_id',
    how='left'
)

# Adicionar as informações dos produtos
df_completo = df_completo.merge(
    df_produtos[['produto_id', 'categoria']],
    on='produto_id',
    how='left'
)

# Coluna Faturamento
df_completo['faturamento'] = (
    df_completo['quantidade'] * df_completo['preco_unitario']
)

# Coluna mes
df_completo['mes'] =df_completo['data_pedido'].dt.to_period('M')

tabela_mensal=pd.pivot_table(
    df_completo,
    index='mes',
    values='faturamento',
    aggfunc='sum'
)

tabela_mensal=tabela_mensal.sort_values(by='faturamento',ascending=False)
display(tabela_mensal)

# Mes com menor faturamento
display(tabela_mensal.tail(1))

```

```python
# Calculando a margem porcentual
display(df_produtos.info())
df_produtos['margem_pct']=(
    (df_produtos['preco_lista']-df_produtos['custo_unitario'])
    /df_produtos['preco_lista'])*100
display(df_produtos['margem_pct'])

# Classificar a margem
condicao=[
    df_produtos['margem_pct']<15,
    (df_produtos['margem_pct']>=15) &
    (df_produtos['margem_pct']<30),
    df_produtos['margem_pct']>=30
]

classificacoes=[
    'Margem Baixa',
    'Margem Média',
    'Margem Alta'
]

df_produtos['Classificação']=np.select(condicao,classificacoes,default='Verificar')
display(df_produtos)
```

```python
# Obtendo a coluna quantidade
quantidade=df_itens_pedidos['quantidade'].dropna()

# Media da quantidade
media=np.mean(quantidade)

# Mediana da quantidade
mediana=np.median(quantidade)

# Desvio padrão da quantidade
desvio_padrao=np.std(quantidade)

print(f"Média:{media}\nMediana:{mediana}\nDesvio Padrão:{desvio_padrao}")

#Calcular os percentis

p25=np.percentile(quantidade,25)
p50=np.percentile(quantidade,50)
p75=np.percentile(quantidade,75)
p90=np.percentile(quantidade,90)

print('Percentil 25:', p25)
print('Percentil 50:', p50)
print('Percentil 75:', p75)
print('Percentil 90:', p90)
```

```python
#juntando tabelas 
juncao = df_itens_pedidos.merge(
    df_produtos,
    on='produto_id',
    how='inner',
    validate='many_to_one'
)
# Calcular receita e custo por item
juncao['receita_item'] = (
    juncao['quantidade'] * juncao['preco_unitario']
)

juncao['custo_item'] = (
    juncao['quantidade'] * juncao['custo_unitario']
)


# Somar os totais
receita_total = juncao['receita_item'].sum()
custo_total = juncao['custo_item'].sum()
lucro_total = receita_total - custo_total

print(f"Receita total: R$ {receita_total:,.2f}")
print(f"Custo total:   R$ {custo_total:,.2f}")
print(f"Lucro total:   R$ {lucro_total:,.2f}")
print(f"Linhas após junção: {len(juncao)}")
```
